```python
from dotenv import load_dotenv
from pathlib import Path  
from botocore.exceptions import ClientError
import boto3
import os

envPath = Path.cwd().parent.parent.joinpath("env").joinpath("dev.aws.env")

print(envPath)

if envPath.exists():
    load_dotenv(dotenv_path= envPath, override=True)
    print("env loaded")
else:
    print("Env file missing")


endpoint_url = os.getenv("AWS_URL")
region = os.getenv("REGION")
aws_access_key_id = os.getenv("AWS_ACCESS_KEY_ID")
aws_secret_access_key = os.getenv("AWS_SECRET_ACCESS_KEY")

print(os.getenv("APP_NAME"), endpoint_url, region, aws_access_key_id, aws_secret_access_key)
```

```python
try:
    """Creating Clients"""
    lambda_client =boto3.client("lambda", endpoint_url =endpoint_url,
        region_name=region,
        aws_access_key_id=aws_access_key_id,
        aws_secret_access_key=aws_secret_access_key
    )

    s3_client = boto3.client("s3", endpoint_url =endpoint_url, region_name=region, aws_access_key_id=aws_access_key_id, aws_secret_access_key=aws_secret_access_key)

    print(lambda_client, s3_client)
except ClientError as ce:
    print("===============>>>>>>")
    print(ce)
except Exception as e:
    print("===============>")
    print(e)
```

```python
try:
    bucket_name = "lambda"
    bucket = s3_client.create_bucket(Bucket=bucket_name)

except ClientError as ce:
    if ce.response['Error']['Code'] == "InvalidBucketName":
        print("Invalid Bucket Name")
    else:
        print("===============>>>>>>")
        print(ce)
except Exception as e:
    print("===============>")
    print(e)
```

```python
from docker.api.exec_api import ExecApiMixin
from IPython.core import extensions
from IPython.core import prefilter
from IPython.core import prefilter
""" Upload code file to S3 """
code_file = Path.cwd().parent.joinpath("aws").joinpath("lambda_functions").joinpath("hello_world.py")

print(code_file)
if not code_file.exists():
    print("File not found")
else:
    file_path = str(code_file)
    file_name = file_path.split("\\")[-1]
    try:
        print(f"Uploading {file_name} to bucket {bucket_name} as {file_name}...")
        s3_client.upload_file(code_file, bucket_name, file_name)
        print("File uploaded")
    except ClientError as ce:
        print(ce)
    except ExecApiMixin as e:
        print(e)
```
