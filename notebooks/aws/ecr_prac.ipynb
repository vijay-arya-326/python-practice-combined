```python
from dotenv import load_dotenv
from pathlib import Path  
envPath = Path.cwd().parent.parent.joinpath("env").joinpath("dev.aws.env")

print(envPath)

if envPath.exists():
    load_dotenv(dotenv_path= envPath, override=True)
    print("env loaded")
else:
    print("Env file missing")
```
