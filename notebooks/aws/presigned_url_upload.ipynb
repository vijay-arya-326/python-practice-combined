```python
from dotenv import load_dotenv
from pathlib import Path  
envPath = Path.cwd().parent.parent.joinpath("env").joinpath("dev.aws.env")

print(envPath)

if envPath.exists():
    load_dotenv(dotenv_path= envPath, override=True)
    print("env loaded")
else:
    print("Env file missing")
```

```python
from botocore.exceptions import ClientError
import boto3
import os

endpoint_url = os.getenv("AWS_URL")
region = os.getenv("REGION")
aws_access_key_id = os.getenv("AWS_ACCESS_KEY_ID")
aws_secret_access_key = os.getenv("AWS_SECRET_ACCESS_KEY")

bucket_name = "my-local-bucket"

print(
    os.getenv("APP_NAME"), 
    endpoint_url, 
    region, 
    aws_access_key_id, 
    aws_secret_access_key
    )
```

```python
s3_client = boto3.client("s3",
endpoint_url = endpoint_url,
region_name = region,
aws_access_key_id =aws_access_key_id,
aws_secret_access_key = aws_secret_access_key 
)

s3_client
```

```python
def create_presigned_url(bucket_name, object_name ,expire_time=3600):
    try:
        response = s3_client.generate_presigned_url(
            ClientMethod='put_object',
                Params={
                    'Bucket': bucket_name, 
                    'Key': object_name
                },
                ExpiresIn=expire_time
        )
        return response

    except ClientError as ce:
        print(ce.response["Error"]["Code"])
    except Exception as e:
        print(e)
```

```python
# Usage Example:
bucket = "my-local-bucket"
file_key = ("Generative-AI-and-LLMs-for-Dummies.pdf")
upload_url = create_presigned_url(bucket, file_key)

print(f"Presigned URL:\n{upload_url}")
```
